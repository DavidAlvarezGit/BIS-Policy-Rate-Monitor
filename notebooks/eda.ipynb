## 5. Analyze some countries


```python
COUNTRIES = ["US", "XM", "GB", "JP", "CH"]
START_DATE = "2015-01-01"

selected = df.loc[
    df["country_code"].isin(COUNTRIES)
    & (df["observation_date"] >= START_DATE)
].copy()

selected.groupby(["country_code", "country_name"])["observation_value"].agg(
    ["count", "min", "max"]
)
```

```python
latest = (
    selected.sort_values("observation_date")
    .groupby(["country_code", "country_name"], as_index=False)
    .tail(1)
    [["country_code", "country_name", "observation_date", "observation_value"]]
    .sort_values("country_code")
)

latest
```

## 6. Check monthly rates are equal to daily rates

```python
daily = df.loc[df["frequency"].eq("D")].copy()
monthly = df.loc[df["frequency"].eq("M")].copy()

daily = daily.dropna(subset=["observation_value"])

daily["month"] = daily["observation_date"].dt.to_period("M")
monthly["month"] = monthly["observation_date"].dt.to_period("M")

daily_month_end = (
    daily.sort_values("observation_date")
    .groupby(["country_code", "month"], as_index=False)
    .tail(1)
)

comparison = monthly.merge(
    daily_month_end[
        [
            "country_code",
            "month",
            "observation_date",
            "observation_value",
        ]
    ],
    on=["country_code", "month"],
    suffixes=("_monthly", "_daily"),
)

comparison["status"] = "different"

comparison.loc[
    comparison["observation_value_monthly"].isna(),
    "status",
] = "monthly_missing"

comparison.loc[
    comparison["observation_value_monthly"].notna()
    & (
        comparison["observation_value_monthly"]
        == comparison["observation_value_daily"]
    ),
    "status",
] = "same"

comparison[
    [
        "country_code",
        "month",
        "observation_value_monthly",
        "observation_value_daily",
        "status",
    ]
].head(10)
```
